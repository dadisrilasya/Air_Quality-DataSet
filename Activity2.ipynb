```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset and strip whitespace from columns/string values
df = pd.read_csv('Air_Quality.csv')
df.columns = df.columns.str.strip()
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()

# Standardize location names to title case to avoid duplicate categories (e.g., 'city' vs 'City')
if 'Location' in df.columns:
    df['Location'] = df['Location'].str.title()

#--- DATA PREPARATION & AGGREGATION ---
# Filter data for just the City
city_center_data = df[df['Location'] == 'City']

# Group data by location to find average PM2.5 pollution levels
avg_pollution = df.groupby('Location')[['PM2_5']].mean()

# --- THE 5 REQUIRED CHARTS ---
# Chart 1: Line Chart (Tracking PM2.5 levels over time at City)
plt.figure(figsize=(8,4))
plt.plot(city_center_data['Date'], city_center_data['PM2_5'], marker='o', color='red')
plt.title('Chart 1: PM2.5 Level Trend at City')
plt.xlabel('Date')
plt.ylabel('PM2_5')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Chart 3: Scatter Plot (Distribution of PM2.5 Levels over Dates)
plt.figure(figsize=(8, 4))
plt.scatter(df['Date'], df['PM2_5'], color='purple', s=50)
plt.title('Chart 3: PM2.5 Level Distribution by Date')
plt.xlabel('Date')
plt.ylabel('PM2_5')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/Air_Quality.csv')

#Chart4: Histogram
plt.figure(figsize=(8, 4))
plt.hist(df['PM2_5'], bins=20, color='green', edgecolor='black')
plt.title('Chart 4: PM2_5 Level Distribution')
plt.xlabel('PM2_5')
plt.ylabel('Frequency')
plt.show()
```

```python
#char5:Horizantal Bar chart
plt.figure(figsize=(8,4))
df.groupby('Location')['PM2_5'].max().plot(kind='barh', color='blue')
plt.title('Chart 5: Highest PM2_5 Levels by Location')
plt.xlabel('PM2_5 Level')
plt.ylabel('Location')
plt.show()
```

```python
# Chart 6: Vertical Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2_5'].max().plot(kind='bar', color='teal')
plt.title('Chart 5: Maximum PM2.5 Recorded per Location')
plt.xlabel('Location')
plt.ylabel('Peak PM2_5 Level')
plt.xticks(rotation=0)
plt.show()
```
